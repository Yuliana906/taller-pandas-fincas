## 6. Filtrado categórico por lista (isin)
**Enunciado:** Seleccionar los predios de los municipios Anserma y Risaralda.

```python
# Filtrar por lista de valores (isin)
df_fincas[df_fincas["Municipio"].isin(["Anserma", "Risaralda"])]
```

## 7. Filtrado por intervalo numérico (between)
**Enunciado:** Aislar los lotes con Area_Hectareas entre 3.0 y 8.0 inclusive.

```python
# Filtrar por área entre 3.0 y 8.0 hectáreas (inclusive)
df_fincas[df_fincas["Area_Hectareas"].between(3.0, 8.0)]
```

## 8. Consultas directas con .query()
**Enunciado:** Extraer fincas con Certificacion == 'Orgánico' y Kilos_Recolectados > 5000.

```python
# Filtrar por una query
df_fincas.query("Certificacion == 'Orgánico' and Kilos_Recolectados > 5000")
```

## 9. Filtrado compuesto con métodos de texto
**Enunciado:** Fincas cuyo Metodo_Beneficio inicie con 'Lavado' y cuya Variedad sea 'Geisha'.

```python
# Filtrar por condiciones y conectores
lavado = df_fincas["Metodo_Beneficio"].str.startswith("Lavado")
geisha = df_fincas["Variedad"] == "Geisha"

filtro = lavado & geisha
df_fincas[filtro]
```

## 10. Imputación de valores faltantes
**Enunciado:** Reemplazar NaN de Area_Hectareas por su promedio y los de Kilos_Recolectados por 0.0 con .fillna().

```python
# Contar valores nulos por columna
df_fincas.isna().sum()
```

```python
# Rellenar valores faltantes
df_fincas = df_fincas.fillna(
    {
        "Area_Hectareas": df_fincas["Area_Hectareas"].mean(),
        "Kilos_Recolectados": 0.0
    }
)
df_fincas.isna().sum()
```

## 11. Transformación y creación de columnas con .apply()
**Enunciado:** Crear una función de rendimiento por hectárea y aplicarla para crear Rendimiento_Kilos_Ha.

```python
# Transformación de datos
def rendimiento(fila):
    return fila["Kilos_Recolectados"] / fila["Area_Hectareas"]

df_fincas["Rendimiento_Kilos_Ha"] = df_fincas.apply(rendimiento, axis=1)
df_fincas.head()
```

## 12. Renombrado y reordenamiento de atributos
**Enunciado:** Renombrar Metodo_Beneficio → Beneficio y Altitud_MSNM → Altitud; poner ID_Finca y Rendimiento_Kilos_Ha al inicio.

```python
# Renombrar columnas
df_fincas = df_fincas.rename(
    columns = {
        "Metodo_Beneficio": "Beneficio",
        "Altitud_MSNM": "Altitud"
    }
)
df_fincas.head()
```

```python
# Reordenar columnas
columnas_ordenadas = [
    "ID_Finca", "Rendimiento_Kilos_Ha", "Municipio", "Variedad", "Beneficio",
    "Area_Hectareas", "Kilos_Recolectados", "Altitud", "Certificacion"
]
df_fincas = df_fincas[columnas_ordenadas]
df_fincas.head()
```

## 13. Agrupación y métricas de producción por municipio
**Enunciado:** Agrupar por Municipio y calcular promedio de Altitud y suma de Kilos_Recolectados con .agg().

```python
# Agrupar los datos por municipio
grouped = df_fincas.groupby("Municipio")

# Promedio de altitud y suma de kilos por municipio
aggregated_data = grouped.agg(
    {
        "Altitud": "mean",
        "Kilos_Recolectados": "sum"
    }
)
aggregated_data
```

## 14. Dispersión con función personalizada
**Enunciado:** Crear rango(series) = max - min y aplicarla con .agg() sobre Area_Hectareas y Kilos_Recolectados agrupando por Variedad.

```python
# Definir una función de agregación personalizada
def rango(series):
    return series.max() - series.min()

# Agrupar por variedad y aplicar la función
grouped_variedad = df_fincas.groupby("Variedad")
aggregated_data_custom = grouped_variedad.agg(
    {
        "Area_Hectareas": rango,
        "Kilos_Recolectados": rango
    }
)
aggregated_data_custom
```

## 15. Agrupamiento multinivel
**Enunciado:** Agrupar por Municipio y Variedad, con promedio de Altitud y suma de Kilos_Recolectados.

```python
# Agrupar por municipio y variedad
grouped_multi = df_fincas.groupby(["Municipio", "Variedad"])

# Promedio de altitud y suma de kilos por cada combinación
aggregated_multi = grouped_multi.agg(
    {
        "Altitud": "mean",
        "Kilos_Recolectados": "sum"
    }
)
aggregated_multi
```
